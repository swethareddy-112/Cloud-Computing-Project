```python
import pandas as pd
import os

# File to store tasks
FILE = "tasks.csv"

# Create file if it doesn't exist
if not os.path.exists(FILE):
    df = pd.DataFrame(columns=["Task", "Status"])
    df.to_csv(FILE, index=False)

# Function to load tasks
def load_tasks():
    return pd.read_csv(FILE)

# Function to save tasks
def save_tasks(df):
    df.to_csv(FILE, index=False)

# Menu
while True:
    print("\n===== Cloud Task Management System =====")
    print("1. Add Task")
    print("2. View Tasks")
    print("3. Mark Task as Completed")
    print("4. Delete Task")
    print("5. Exit")

    choice = input("Enter your choice: ")

    df = load_tasks()

    if choice == "1":
        task = input("Enter task: ")
        df.loc[len(df)] = [task, "Pending"]
        save_tasks(df)
        print("Task Added Successfully!")

    elif choice == "2":
        if df.empty:
            print("No tasks available.")
        else:
            print(df)

    elif choice == "3":
        if df.empty:
            print("No tasks available.")
        else:
            print(df)
            index = int(input("Enter task number (index): "))
            if index in df.index:
                df.at[index, "Status"] = "Completed"
                save_tasks(df)
                print("Task Completed!")
            else:
                print("Invalid index.")

    elif choice == "4":
        if df.empty:
            print("No tasks available.")
        else:
            print(df)
            index = int(input("Enter task number (index): "))
            if index in df.index:
                df = df.drop(index).reset_index(drop=True)
                save_tasks(df)
                print("Task Deleted!")
            else:
                print("Invalid index.")

    elif choice == "5":
        print("Thank you!")
        break

    else:
        print("Invalid choice.")
```
